```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv('../data/raw/train.csv.zip')
print(df.shape) # returns dimensions (rowsxcol)
print(df.head(10)) # first 10 rows
```

```python
# label distribution
# what percentage of comments are toxic vs clean

label_cols = ['toxic','severe_toxic','obscene','threat','insult','identity_hate']
print(df[label_cols].sum())
```

```python
print(df[label_cols].mean() * 100) 
```

```python
print(df[label_cols].corr()) # how many times two labels appear on same comment
```
